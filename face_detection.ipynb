```python
import cv2
from ultralytics import YOLO
```

### 1_reading image and show it using cv2

```python
img_path = "./images/pedestrain_car.jpg"
img = cv2.imread(img_path)
cv2.imshow("window1", img)
cv2.moveWindow("window1",250, 250)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

### 2_loading model and process image 

```python
model = YOLO("yolov8n.pt")
```

```python
result = model(img)
```

```python
cv2.imshow("window1", result[0].plot())
cv2.moveWindow("window1", 250, 250)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```
