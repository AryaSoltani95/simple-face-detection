```python
import cv2
from ultralytics import YOLO
```

### 1_reading image and show it using cv2

```python
img_path = "./images/pedestrain_car.jpg"
img = cv2.imread(img_path)
cv2.imshow("window1", img)
cv2.moveWindow("window1",0, 0)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

### 2_loading model and process image 

```python
model = YOLO("yolov8n.pt")
```

```python
object_detection_result = model(img)
cv2.imshow("window1", object_detection_result[0].plot())
cv2.moveWindow("window1", 0, 0)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

### 3_load comunity version of face detection 

```python
face_model = YOLO("face_detection_models/yolov12m-face.pt")
```

```python
face_detection_result = face_model(img)
cv2.imshow("window1", face_detection_result[0].plot())
cv2.moveWindow("window1", 0, 0)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

### 4_using both models for face and object detection 

```python
img_path = "./images/pedestrain_car.jpg"
img = cv2.imread(img_path)
cv2.imshow("window1", img)
cv2.moveWindow("window1", 0, 0)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
object_model = YOLO("yolov8n.pt")
face_model = YOLO("./face_detection_models/yolov12m-face.pt")
```

```python
object_model_result = object_model(img)
face_model_result = face_model(img)
```

```python
detected_objects = object_model_result[0].plot()
# detected_faces = face_model_result[0].plot() 
detected_face_objects = face_model_result[0].plot(img=detected_objects)
```

```python
cv2.imshow("window1", detected_face_objects)
cv2.moveWindow("window1", 0, 0)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```
